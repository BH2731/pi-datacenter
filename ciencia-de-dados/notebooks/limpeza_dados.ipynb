# Etapa 2 — Tratamento de células nulas

As colunas do Grupo 12 são:
`consumo_total_kwh`, `energia_gpus_kwh`, `eficiencia_energetica_pct` e `status_eficiencia`.

Nas colunas numéricas será usada a média. Na coluna categórica será usada a moda.

```python
dados[['consumo_total_kwh', 'energia_gpus_kwh', 'eficiencia_energetica_pct', 'status_eficiencia']].isna().sum()
```

```python
mediaConsumo = dados['consumo_total_kwh'].mean()
mediaEnergia = dados['energia_gpus_kwh'].mean()
mediaEficiencia = dados['eficiencia_energetica_pct'].mean()
modaStatus = dados['status_eficiencia'].mode()[0]

dados.fillna({
    'consumo_total_kwh': mediaConsumo,
    'energia_gpus_kwh': mediaEnergia,
    'eficiencia_energetica_pct': mediaEficiencia,
    'status_eficiencia': modaStatus
}, inplace=True)
```

```python
dados[['consumo_total_kwh', 'energia_gpus_kwh', 'eficiencia_energetica_pct', 'status_eficiencia']].isna().sum()
```

# Etapa 3 — Tratamento de valores negativos impossíveis

As colunas são:
`energia_gpus_kwh`, `energia_refrigeracao_kwh` e `consumo_agua_l`.

```python
mascaraEnergia = dados['energia_gpus_kwh'] < 0
dados[mascaraEnergia]
```

```python
mascaraRefrigeracao = dados['energia_refrigeracao_kwh'] < 0
dados[mascaraRefrigeracao]
```

```python
mascaraAgua = dados['consumo_agua_l'] < 0
dados[mascaraAgua]
```

```python
mediaEnergia = dados[dados['energia_gpus_kwh'] >= 0]['energia_gpus_kwh'].mean()
mediaRefrigeracao = dados[dados['energia_refrigeracao_kwh'] >= 0]['energia_refrigeracao_kwh'].mean()
mediaAgua = dados[dados['consumo_agua_l'] >= 0]['consumo_agua_l'].mean()

dados.loc[mascaraEnergia, 'energia_gpus_kwh'] = mediaEnergia
dados.loc[mascaraRefrigeracao, 'energia_refrigeracao_kwh'] = mediaRefrigeracao
dados.loc[mascaraAgua, 'consumo_agua_l'] = mediaAgua
```

# Etapa 4 — Tratamento de percentuais inválidos

Os valores devem ficar entre 0 e 100.

```python
mascaraEnergetica = (dados['eficiencia_energetica_pct'] < 0) | (dados['eficiencia_energetica_pct'] > 100)
dados[mascaraEnergetica]
```

```python
mascaraHidrica = (dados['eficiencia_hidrica_pct'] < 0) | (dados['eficiencia_hidrica_pct'] > 100)
dados[mascaraHidrica]
```

```python
mediaEnergetica = dados[(dados['eficiencia_energetica_pct'] >= 0) & (dados['eficiencia_energetica_pct'] <= 100)]['eficiencia_energetica_pct'].mean()
mediaHidrica = dados[(dados['eficiencia_hidrica_pct'] >= 0) & (dados['eficiencia_hidrica_pct'] <= 100)]['eficiencia_hidrica_pct'].mean()

dados.loc[mascaraEnergetica, 'eficiencia_energetica_pct'] = mediaEnergetica
dados.loc[mascaraHidrica, 'eficiencia_hidrica_pct'] = mediaHidrica
```

# Etapa 9 — Agrupamento e análise dos dados

O agrupamento será feito por `zona_cluster_ai` e `turno`.

```python
agrupamento = dados.groupby(['zona_cluster_ai', 'turno']).agg(
    media_consumo_total=('consumo_total_kwh', 'mean'),
    media_energia_gpus=('energia_gpus_kwh', 'mean'),
    quantidade=('id_registro', 'count')
)

agrupamento
```

# Etapa 10 — Interpretação dos resultados

Abaixo são identificados dois resultados importantes do agrupamento.

```python
resultadoConsumo = agrupamento.sort_values('media_consumo_total', ascending=False).reset_index()
resultadoConsumo.head(1)
```

```python
resultadoGpu = agrupamento.sort_values('media_energia_gpus', ascending=False).reset_index()
resultadoGpu.head(1)
```

```python
maiorConsumo = resultadoConsumo.iloc[0]
maiorGpu = resultadoGpu.iloc[0]
```
